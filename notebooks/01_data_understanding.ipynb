```python
import re

def normalize(title):
    title = str(title).lower()
    title = re.sub(r"\(.*?\)", "", title)
    title = re.sub(r"\[.*?\]", "", title)
    title = re.sub(r"[^a-z0-9 ]", "", title)
    return title.strip()

hindi_df["clean_title"] = hindi_df["song_name"].apply(normalize)
lyrics_df["clean_title"] = lyrics_df["Song Title"].apply(normalize)
haryanvi_df["clean_title"] = haryanvi_df["song_name"].apply(normalize)
old_songs_df["clean_title"] = old_songs_df["song_name"].apply(normalize)
punjabi_df["clean_title"] = punjabi_df["song_name"].apply(normalize)
```

```python
import pandas as pd

def standardize_dtypes(df: pd.DataFrame, dataset_name: str = "") -> pd.DataFrame:

    str_cols = ['song_name', 'singer', 'language']
    for col in str_cols:
        df[col] = df[col].astype(str).str.strip().str.lower()

    df['released_date'] = pd.to_datetime(df['released_date'], errors='coerce')

    float_cols = [
        'danceability', 'acousticness', 'energy',
        'liveness', 'loudness', 'speechiness',
        'tempo', 'Valence'
    ]
    for col in float_cols:
        df[col] = pd.to_numeric(df[col], errors='coerce').astype(float)

    int_cols = ['mode', 'key', 'time_signature', 'popularity']
    for col in int_cols:
        df[col] = pd.to_numeric(df[col], errors='coerce').astype('Int64')  # Int64 handles NaN, int doesn't

    def parse_duration(val):
        try:
            if isinstance(val, str) and ':' in val:
                parts = val.strip().split(':')
                minutes, seconds = int(parts[0]), int(parts[1])
                return (minutes * 60 + seconds) * 1000
            return int(float(val))
        except:
            return pd.NA
    df['duration'] = df['duration'].apply(parse_duration).astype('Int64')
    df['Stream'] = pd.to_numeric(df['Stream'], errors='coerce').astype('Int64')

    print(f"\n{'='*50}")
    print(f"Dataset: {dataset_name}")
    print(f"Shape: {df.shape}")
    print(f"Null counts:\n{df.isnull().sum()[df.isnull().sum() > 0]}")
    print(f"{'='*50}")

    return df
```
